```python
import os
import glob
from IPython.display import Image, display, HTML

# Define the base directory where plots are saved
base_plot_dir = "training_plots"

# --- 1. Find the most recent training run directory ---
# List all directories within base_plot_dir
run_dirs = [d for d in glob.glob(os.path.join(base_plot_dir, "run_*")) if os.path.isdir(d)]

if not run_dirs:
    print(f"No training run directories found in '{base_plot_dir}'. Please ensure training has run.")
else:
    # Sort them by name (which includes the timestamp, so lexicographical sort works for dates/times)
    run_dirs.sort()
    latest_run_dir = run_dirs[-1] # The last one will be the most recent
    print(f"Displaying plots from the latest run: '{latest_run_dir}'")

    # --- 2. List all plot image files in that directory ---
    plot_files = sorted(glob.glob(os.path.join(latest_run_dir, "*.png")))

    if not plot_files:
        print(f"No plot images found in '{latest_run_dir}'.")
    else:
        print(f"Found {len(plot_files)} plots. Displaying a selection:")

        # --- 3. Display a selection of images ---
        # Option A: Display the first few and the last few plots
        num_to_display = 5 # Adjust as needed
        if len(plot_files) <= num_to_display:
            files_to_show = plot_files
        else:
            files_to_show = plot_files[:num_to_display//2] + plot_files[-num_to_display//2:]
            print(f"(Showing {len(files_to_show)} plots: first {num_to_display//2} and last {num_to_display//2})")

        for f_path in files_to_show:
            print(f"\n--- {os.path.basename(f_path)} ---")
            display(Image(filename=f_path))

        # Option B: Or, if you want to display ALL of them (be careful with many epochs!)
        # print("\n--- Displaying ALL plots (can be slow if many epochs) ---")
        # for f_path in plot_files:
        #     print(f"\n--- {os.path.basename(f_path)} ---")
        #     display(Image(filename=f_path))

        # Option C: Just display the very last plot
        # print(f"\n--- Displaying the final plot: {os.path.basename(plot_files[-1])} ---")
        # display(Image(filename=plot_files[-1]))
```
