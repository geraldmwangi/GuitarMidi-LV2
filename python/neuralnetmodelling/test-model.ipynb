```python
from guitarchordanalyzer import GuitarChordAnalyzer, CHORD_FORMULAS


# ============================================================
# USAGE EXAMPLES — realistic guitar voicings
# ============================================================
"""
Comprehensive test suite for GuitarChordAnalyzer.
Generates 2 randomized-root examples for every chord type in CHORD_FORMULAS,
builds actual MIDI note sets, and verifies the analyzer correctly identifies
both the chord and its suffix (short and long form).
"""

import random

NOTE_NAMES = ['C', 'C#', 'D', 'D#', 'E', 'F', 'F#', 'G', 'G#', 'A', 'A#', 'B']

# Reasonable guitar register to build test voicings in (E2 to E5-ish)
LOW_BOUND = 40   # E2
HIGH_BOUND = 76  # E5

def build_voicing(root_pc, intervals, base_octave_midi=48):
    """
    Build a MIDI note list from a root pitch class + interval formula.
    Randomizes octave placement slightly and sometimes doubles the root,
    to simulate realistic guitar voicings.
    """
    root_midi = base_octave_midi + root_pc
    # keep root in a sane low-ish guitar range
    while root_midi < LOW_BOUND:
        root_midi += 12
    while root_midi > 55:  # keep root low, chord tones stack above
        root_midi -= 12

    notes = [root_midi]
    for iv in intervals[1:]:
        notes.append(root_midi + iv)

    # 50% chance: double the root an octave up (common on guitar)
    if random.random() < 0.5:
        notes.append(root_midi + 12)

    # 30% chance: double the 5th if present (common open-chord trait)
    if len(intervals) > 2 and random.random() < 0.3:
        fifth_interval = intervals[2] if len(intervals) > 2 else None
        if fifth_interval:
            notes.append(root_midi + fifth_interval + 12)
    # clamp to max 6 notes (guitar strings)
    notes = notes[:6]
    # keep everything under HIGH_BOUND
    notes = [n for n in notes if n <= HIGH_BOUND]
    return sorted(set(notes))


def run_full_test_suite():
    random.seed()  # uses system time -> different roots each run
    print(f"{'CHORD TYPE':16s} {'ROOT':4s} {'MIDI NOTES':30s} {'SHORT':10s} {'LONG':22s} {'LABEL':10s}")
    print("-" * 100)

    for chord_name, intervals in CHORD_FORMULAS.items():
        for example_num in range(2):
            root_pc = random.randint(0, 11)
            root_name = NOTE_NAMES[root_pc]
            notes = build_voicing(root_pc, intervals)

            analyzer = GuitarChordAnalyzer(notes)
            short_result = analyzer.get_chord_suffix(style='short')
            long_result = analyzer.get_chord_suffix(style='long')

            print(f"{chord_name:16s} {root_name:4s} {str(notes):30s} "
                  f"{str(short_result['suffix']):10s} {str(long_result['suffix']):22s} "
                  f"{str(short_result['full_label']):10s}")


if __name__ == "__main__":
    run_full_test_suite()

    notes=[40,47,52]
    analyzer = GuitarChordAnalyzer(notes)
    analyzer.analyze()
    print(analyzer.get_chord_suffix(style='long'))
    print(analyzer.get_chord_suffix(style='short'))
```
